# T1

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------- 1. 数据准备 ----------------------
your_data = np.load("T1_data.npy")  # 从文件加载数据

sorted_data = np.sort(your_data)
n_points = len(sorted_data)
cumulative_y = np.arange(1, n_points + 1) / n_points

# 扩展数据，让阶梯线连接到坐标轴
sorted_data_ext = np.concatenate([[sorted_data[0]], sorted_data, [sorted_data[-1]]])
cumulative_y_ext = np.concatenate([[0.0], cumulative_y, [1.0]])

# 计算统计量
min_val = sorted_data.min()
max_val = sorted_data.max()
mean_val = np.mean(your_data)
median_val = np.median(your_data)
# ---------------------- 2. 绘图配置（🔹 增大字体大小） ----------------------
plt.rcParams["font.family"] = "Arial"
plt.rcParams["axes.linewidth"] = 1.2
plt.rcParams["xtick.labelsize"] = 20  # 🔹 横坐标刻度：14→20
plt.rcParams["ytick.labelsize"] = 20  # 🔹 纵坐标刻度：14→20
plt.rcParams["axes.labelsize"] = 22   # 🔹 坐标轴标签：16→22

# 小图尺寸（和参考图比例一致）
fig, ax = plt.subplots(figsize=(4, 5), dpi=120)

# ---------------------- 3. 绘制红色阶梯线 ----------------------
ax.step(
    sorted_data_ext, cumulative_y_ext,
    where='post',
    color='#DB312D',
    linewidth=3,
    alpha=0.9
)

# ---------------------- 4. 绘制中位数虚线 + 文字标注（🔹 增大中位数文字） ----------------------
ax.axvline(
    median_val,
    color='#4B74B2',
    linestyle='--',
    linewidth=2.5,
    alpha=0.8
)

ax.text(
    median_val -0.5,  # 🔹 微调位置（字体变大，偏移量适当增加）
    0.5,
    f"Median: {median_val:.2f}",
    color='#4B74B2',
    va='center',
    ha='right',
    fontsize=18,     # 🔹 中位数文字：12→18
    fontweight='medium'
)

# ---------------------- 5. 坐标轴与美化 ----------------------
ax.set_xlim(min_val - 4, max_val + 4)
ax.set_xticks([min_val, max_val])
ax.set_xticklabels([f"{min_val:.2f}", f"{max_val:.2f}"])

ax.set_ylim(0, 1)
ax.set_yticks([0, 0.25, 0.5, 0.75, 1.0])

ax.set_xlabel("$T_1$ ($\mu$s)", fontweight='medium')
ax.set_ylabel("Cumulative Probability", fontweight='medium')

# 美化边框（可选，小图建议保留边框更清晰）
# ax.spines["top"].set_visible(False)
# ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
print(np.median(your_data))
```
