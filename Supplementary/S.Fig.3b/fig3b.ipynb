```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------- 1. 可调节参数（集中控制字体/样式） ----------------------
# 字体大小（直接修改这里即可调整）
XTICK_SIZE = 30       # 横坐标刻度字体大小
YTICK_SIZE = 30       # 纵坐标刻度字体大小
AX_LABEL_SIZE = 30    # 坐标轴标签字体大小
LEGEND_SIZE = 25      # 图例字体大小

# 点大小/线宽
POINT_SIZE = 250       # 数据点大小
LINE_WIDTH = 3        # 连线粗细
AXIS_LINE_WIDTH = 1.2 # 坐标轴边框粗细

# ---------------------- 2. 数据准备（替换为你的两组20个温度数据，单位：mK） ----------------------
temp_group1 = np.load("thermal_noise_removed_data.py.npy")  # 第一组（优化后）
temp_group2 = np.load("direct_measure_Test_data.py.npy") # 第二组（原始）

# 横坐标：仅显示1-20（去掉N），位置对应0-19
x_labels = [str(i+1) for i in range(20)]  # 改为1,2,...,20
x_pos = np.arange(len(x_labels))

# ---------------------- 3. 绘图配置（PRL学术风格） ----------------------
plt.rcParams["font.family"] = "Arial"
plt.rcParams["axes.linewidth"] = AXIS_LINE_WIDTH
plt.rcParams["xtick.labelsize"] = XTICK_SIZE
plt.rcParams["ytick.labelsize"] = YTICK_SIZE
plt.rcParams["axes.labelsize"] = AX_LABEL_SIZE
plt.rcParams["legend.fontsize"] = LEGEND_SIZE

# 创建画布（宽高比适配20个横坐标）
fig, ax = plt.subplots(figsize=(20, 10), dpi=600)

# ---------------------- 4. 绘制哑铃图核心 ----------------------
# 绘制两组数据的连线（灰色，细实线）
for i in range(len(x_pos)):
    ax.plot([x_pos[i], x_pos[i]], [temp_group2[i], temp_group1[i]], 
            color="#CCCCCC", linewidth=LINE_WIDTH, alpha=0.5, zorder=1)

# 第一组数据点（蓝色实心圆）
ax.scatter(x_pos, temp_group1, color="#1f77b4", s=POINT_SIZE, 
           edgecolor="white", linewidth=1.5, label="Remove Thermal noise", zorder=3)

# 第二组数据点（红色实心方）
ax.scatter(x_pos, temp_group2, color="#d62728", s=POINT_SIZE, marker="s",
           edgecolor="white", linewidth=1.5, label="Direct Measurement", zorder=2)

# ---------------------- 5. 坐标轴与标注美化 ----------------------
# 横坐标：1-20，不倾斜，居中对齐
ax.set_xticks(x_pos)
ax.set_xticklabels(x_labels, rotation=0, ha="center")  # rotation=0 取消倾斜

# 坐标轴标签
ax.set_xlabel("Number of Qubits (N)", fontweight="bold")  # X轴新标签
ax.set_ylabel("$T_{est}$ (mK)", fontweight="bold")

# 纵坐标范围（上下各留5%留白）
y_min = min(np.min(temp_group1), np.min(temp_group2)) * 0.95
y_max = max(np.max(temp_group1), np.max(temp_group2)) * 1.05
ax.set_ylim(y_min, y_max)

# 隐藏上/右边框
# ax.spines["top"].set_visible(False)
# ax.spines["right"].set_visible(False)

# 图例（右上角，无背景）
ax.legend(loc="upper right", frameon=False)

# 纵坐标网格线（辅助读数）
ax.grid(axis="y", color="#EEEEEE", linestyle="-", linewidth=1, alpha=0.8)

# 紧凑布局
plt.tight_layout()

plt.show()
```
